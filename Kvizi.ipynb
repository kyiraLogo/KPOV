1. **Katere naslove uporablja povezavna plast?**
    - MAC naslove.

2. **Ali lahko stanje omrežnega (IEEE 802.3) stikala nadziramo brez SNMP?**
    - Da, ampak je zaradi velikega števila orodij, ki ga podpirajo, SNMP bolj prikladen.

3. **Ali lahko na enem računalniku poganjamo več SNMP agentov? Če da, kako?**
    - Običajno na sistemu teče en program, ki se obnaša kot več SNMP agentov skupaj.

4. **Ali se lahko računalnik z UEFI zažene prek omrežja?**
    - Da, ali z uporabo CSM in razširitvenim ROM, ali z uporabo EFI razširitve, ki jo priskrbi izdelovalec mrežnega vmesnika

5. **Omenjali smo, da je ena od glavnih novosti SNMPv3 skrb za varnost. Kako je izvedena zaščita pred ponavljanjem?**
    - Z uporabo kombinacije ID pošiljatelja, števila zagonov in časa po zadnjem zagonu.

6. **Internetni ponudnik uporabniku A zagotavlja statični IP naslov. Ali je možno, da ima A na svojem usmerjevalniku nastavljen DHCP način pridobivanja internetnega naslova?**
    - Da, saj ponudnik lahko nastavi DHCP tako, da A-ju vedno dodeli isti IP naslov (z uporabo MAC naslova modema/prehoda).


7. **Kaj je glavni razlog, da BOOTP in TFTP na transportni plasti uporabljata UDP in ne TCP (en odgovor)?**
    - Želimo majhno kompleksnost.

8. **Naš prijatelj Peter Zmeda je ugotovil, da njegov računalnik uporablja IP naslov 169.254.10.20. Kaj lahko sklepa o omrežju v katerega je priključen.**
    - V omrežju ni delujočega DHCP strežnika.

1. **Ali DHCP protokol uporablja razpošiljevalne naslove?**
    - Da, ampak le na IPv6.

2. **Ali se multicast paketi usmerjajo?**
    - Da. Saj z tem preprečujemo poplavljanje

3. **Vir toka podatkov (SSRC) pošilja podatke sprejemniku. Kako lahko sprejemnik ugotovi kaj je pošiljatelj (kamera, mikrofon, ...)?**
    - Z uporabo NAME polja v RTCP protokolu.

4. **Ali lahko z uporabo razpošiljanja prenašamo celotne datoteke?**
    - Da, multicast se lahko uporablja za prenos celotnih datotek, a ker je osnovno IP multicast nezanesljiv, je za resne prenose potrebno uporabiti dodatne protokole za zanesljivost in nadzor napak.

5. **Kako lahko ugotovimo, ali je katera koli naprava na omrežju prijavljena na neko konkretno razpošiljevalno skupino (multicast group)?**
    - Z uporabo IGMPv1, IGMPv2 ali IGMPv3 poizvedbe.

6. **Kam bo IPv6 paket s ciljnim naslovom FF02::D dostavljen?**
    - Vsem NTP strežnikom na lokalnem omrežju.

7. **Ali stikalo lahko pridobi informacije o skupinah razpošiljanja in se tako izogne razpošiljanju paketov vsem članom podomrežja?**
    - Da, s pomočjo IGMP vohljanja.
