# Upravljanje omrežja preko SNMP

| Ukaz | Kaj naredi | Primer |
|------|------------|--------|
| `snmpwalk` | Zbira informacije iz naprav prek SNMP; izpiše celotno MIB drevo. Odjemalec pošilja zahteve, ne določa, katere podatke agent razkriva. Lahko spremeni verzijo SNMP, community string in naslov naprave. | `snmpwalk -v2c -c public 192.168.1.1 1.3.6.1.2.1.1` <br> *//192.168.1.1 = omrežje, 1.3.6.1.2.1.1 = MIB* |
| `snmpget` | Preveri specifične vrednosti, npr. koliko prostora je kjer | `snmpget -v1 -c studentje localhost .1.3.6.1.4.1.2021.9.1.9.1` <br> *OID lokalnega diska* |
| `snmpset` | Nastavlja vrednosti in upravlja agente SNMP | `snmpset -v1 -c studentje localhost .1.3.6.1.4.1.2021.9.1.9.1 s "NamenUpravljanja"` |

## Pogosti sistemski nalagalniki

- **GNU**:
    - GRand Unified Bootloader (GRUB) – zmogljiv odprtokodni nalagalnik iz projekta GNU, ki podpira več operacijskih sistemov in konfiguracij.

- **systemd-boot**:
    - preprost  zagonski upravitelj (prej znan kot gummiboot), vključen v systemd, ki omogoča izbiro med nameščenimi sistemi in urejanje parametrov jedra.

- **rEFInd**:
    - grafični zagonski upravitelj za UEFI sisteme, namenjen enostavnemu izbiranju med več operacijskimi sistemi; nastal je kot razširitev opuščenega projekta rEFIt.

- **Limine**:
    - sodoben, večplatformni zagonski nalagalnik z podporo za BIOS in UEFI, razvit kot referenčna implementacija Limine zagonskega protokola; omogoča zagon Linuxa in veriženje (chainload) drugih nalagalnikov.
    
- **SYSLINUX**:
    - zbirka lahkih zagonskih nalagalnikov, prvotno namenjenih BIOS/MBR sistemom. Vključuje več specializiranih različic, npr. ISOLINUX za zagonske CD/DVD medije (ISO 9660), PXELINUX za omrežni zagon prek PXE protokola, in EXTLINUX/EFILINUX za zagon z različnih datotečnih sistemov (tudi UEFI podpora v EFILINUX modulu).
    
- **LILO**:
    - starejši Linux zagonski nalagalnik za BIOS sisteme; zgodovinsko je bil privzet v mnogih distribucijah pred prehodom na GRUB (razvoj LILO je bil zaključen leta 2015).
