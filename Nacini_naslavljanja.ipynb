## Omrežna plast: [Omrezne_plasti](.\Omrezne_Plasti.ipynb)

- **Unicast:**
    - naslavljanje posameznega omrežnega vmesnika
    - pošiljanje enemu ciljnemu IP naslovu (unikaten v Internetu / lokalnemu omrežju)

- **Multicast:**    
[Multicasting](./Multicasting.ipynb)
    - mejne skupine (e-mail(upravljalec oseba), radijo(upravljalec omrežni protokoli))
    - naslavlnanje skupine omrežnih vmesnikov
    **IPv6:**
     - FF02::1 (vmesnik)
     - FF02::2 (usmerjevalnik)
     - [IPv6_posebni](./Slike/IPv6_posebni.png)

- **Anycast:**
    - je naslov množice vmesnikov, dostava se izvede enemu vmesniku iz te množice

- **Broadcast:**
    - pošlje vsem v omejenem omrežju (ARP poizvedba)
    - V IPv6 nima (namesto tega multicast)
    - naslavljanje vseh prejemnikov“ v podomrežju (npr. iskanje usmerjevalnika ali strežnika, nujno sporočilo); ne dostavlja paketov izven lokalne mreže
