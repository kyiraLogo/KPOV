## DHCP
**Dynamic Host Configuration Protocol**
- Namenjen razdeljevanju IP naslovov. Zapomni si kdo ima kateri naslov, koliko časa.
- DHCP je aplikacija, ki na prenosni plasti uporablja **nepovezavni način – UDP protokol**
- **Odjemalec dobi na uporabo IP naslov za določen čas**
    - Po potečenem času:
    - Pošlje prošno za podaljšanje (če potrebuje)
    - Pošlje zaključek uporabe
    - *V primeru, da ne vrne IP:*
        - DHCP strežnik kaznuje odjemalca,blokira dostop do povezovanja z internetom (komunicira in upravlja usmerjevalnik)

Možna konfiguracija (dhclient): [/etc/dhclient.conf](./Ukazi-Faili.ipynb)
Možna konfiguracija (net/isc-dhcp31-server): [/usr/local/etc/dhcpd.conf](./Ukazi-Faili.ipynb)


## Zahteve
- DHCPDISCOVER: iskanje strežnika
    - **DHCPv4: uporablja broadcast (255.255.255.255) za odkrivanje strežnika, ne uporablja multicast.**
- DHCPOFFER: ponudba odjemalcu
- DHCPREQUEST: odjemalec potrjuje prejete nastavitve; tudi želja po podaljšanju sposoje IP naslova
- DHCPACK, DHCPNAK: strežnikova potrditev/zanikanje odjemalcu
- DHCPDECLINE: odjemalec strežniku, da je IP naslov že v uporabi
- DHCPRELEASE: odjemalec vrača naslov pred potekom
- DHCPINFORM: odjemalec želi samo ostale podatke, naslov že ima

**Pridobitev IP**
- **DHCP prošnja:** *(enkapsulacija)*
    - UDP (doda port in omogoča prenos podatkov) port:67/68
    - IP (doda svoj IP in IP od DHCP) (če ga nima pošle brotcast) *usmerjevalnik lahko upravlja naloga DHCP strež.*
    - Ethernet 802.1 (Doda MAC naslov pošiljatelja in prejemnika)
    - Fizično (kabel, wifi)

- **DHCP odgovor:**
    - pošlje DHCP ACK (njegov IP, naslov prehoda in DNS)
    - naredimo *dekapsulacijo*

**Spletno brskanje**
- Pošljemo HTTP zahtevek:
    - Pridobimo IP domene (DNS)
        - pošljemo enkapsulirano(DNS: UDP-> IP-> Ethernet 802.1-> MAC(Dobimo z ARP))
        - Usmerjevalnik pošlje DNS na IP
        - DNS strežnik dekapsulira
    - vrne IP domene
    - pošljemo HTTP: TCP(vtič, SYN) odgovori: TCP SYNACK (potrdi vzpostavitev povezave) (**3 way hand shake**)
    - pošljemo HTTP zahtevek: TCP, IP diagram
    - dobimo vsebino spletne strani

## Zahteve DHCPv6

- **DHCPv6: uporablja multicast (FF02::1:2) za odkrivanje strežnikov, ker IPv6 nima broadcast mehanizma.**

- SOLICIT: prošnja za nastavitev
- ADVERTISE: oglašanje naslova
- REQUEST: zahteva za nastavitvene parametre
- CONFIRM: preverjanje, ali je naslov, ki ga je dobil odjemalec, še vedno v redu
- RENEW: zahteva za obnovitev
- REBIND: zahteva za ohranitev
- REPLY: odgovor odjemalcu
- RELEASE: sprostitev naslova
- DECLINE: zavrnitev dodeljenega naslova
- RECONFIGURE: strežnik odjemalcu sporoča, naj obnovi nastavitve
- INFORMATION-REQUEST: zahteva za nastavitve brez IP naslova
- RELAY-FORW: prepošiljanje
- RELAY-REPL: potrdilo prepošiljatelju, ki vsebuje odgovor odjemalcu
