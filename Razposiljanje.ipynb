## Razpošiljevalna drevesa

### Internet:
- Internet je graf (drevo)
- Pri grafih je pomembno kako definiramo optimalno pot. 
(Čim manj obremenjijeno onrežje, čim manj hopov,...)    

Poznamo:    
- **Skupno drevo**: -> **RTP**
    - **centralno vozlišče** (roundevu point) skrbi za drevo (ima sve podatke o odjemalcih)
    - če želimo kaj povprašat uprašamo njega
- **Ločeno drevo** (vsako vozlišče ima svojo drevo): -> **SPT**
	- Edini način da povprašamo po omrežju je z poplavljanjem (+ tabele)
	- Paket je autentičen samo če je prišel po najkrajši poti **reverse path look up** (da ne utopimo omrežja)

## PIM-SM   
[PIM_packet](./Slike/PIM_packet.png)    

- PIM-DM: dense-mode, posamezno drevo
- PIM-SM: sparse-mode, skupno drevo, včasih posamezno   
V usmerjevalnikih z unicast (ni odvisen od unicast protokola)   
- Uporablja IP protokol (št. protokola 103) 
Usmerjevalnik: (sporočila) --> unicast, multicast (na 224.0.0.13 (*vsi PIM usmerjevalniki*))    

### Delovanje:
- Vzpostavitev arhitekture
- Pošiljanje podatkov
- Vzdrževenje naročnine

- Različni paketi:
    - [Hello_packet](./Slike/PIM_Hello.png)
    - [Register_packet](./Slike/PIM_Register.png)
    - [Join/Prune_packet](./Slike/PIM_join_prune.png)

- PIM-SM uporablja dva tipa dreves:

    - Shared Tree (RPT – Rendezvous Point Tree)
        - Promet potuje od RP do vseh prejemnikov.
        - Kasneje lahko preklopimo na SPT

    - Source Tree (SPT – Shortest Path Tree)
        - Drevo se gradi od vira prejemnikov
        - Kdor želi promet dobi najkrajšo pot
        - Ko je promet aktiven, se lahko izgradi drevo neposredno od source do prejemnikov (hitrejša pot).


#### Osnovni koncepti

- **Multicast skupina (G)**
- IP naslov (npr. 224.0.0.0 – 239.255.255.255) predstavlja skupino, na katero se lahko naročijo prejemniki.

- **Source (S)**
    - Naprava, ki pošilja multicast promet.

- **Rendezvous Point (RP)**
    - Posebni usmerjevalnik, ki deluje kot centralna točka za začetno usmerjanje multicast prometa.
    - Prejemniki (receivers) se povežejo z RP, da prejmejo promet, ki ga pošilja source.

## PIM Dense Mode (DM)
- Način delovanja: flood-and-prune
- Pošlje multicast promet vsem usmerjevalnikom v omrežju.
- Usmerjevalniki, ki nimajo prejemnikov, kasneje pošljejo prune sporočila, da se promet odstrani z nepotrebnih poti.

Uporaba: ko je večina omrežja zainteresirana za promet (več prejemnikov).   

- **Rendezvous Point: ni potreben.**
- Drevo: SPT-like drevo (direktno od vira do prejemnikov, po poplavljanju). 

Prednosti:   
- Preprosto za implementacijo.

Slabosti:       
- Neoptimalno pri majhnem številu prejemnikov → veliko nepotrebnega prometa.

## Ostali protokoli za usmerjanje
- **MOSPF**:
    - Multicast OSPF
    - ima dodano le posebno obliko paketa, ki oznanja razpošijevalni promet

- **DVMRP**:
    - Distance Vector Multicast Routing Protocol
    - prenaša se ga v IGMP paketih (tip 13)

- **MBONE**:
- povezava omrežij, zmožnih razpošiljevalnega prometa
- sprva znotraj interneta, tvorile so ga delovne postaje z navideznimi povezavami

- IETF ustanovi delovno skupino MBONED z nalogo, da vzpostavi multicast usmerjanje preko celega interneta (razvoj protokola MSDP: Multicast Source Discovery Protocol)
    
### Zgodovina MBONE
- 1995: MBONE vsebuje 901 usmerjevalnikov (uporablja se DVMRP) in je v 20 državah
- 1999: 4178 usmerjevalnikov, uporablja se vse bolj RTP, ponudniki storitev postajajo preobremenjeni
