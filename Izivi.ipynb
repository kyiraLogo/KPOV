- Intel (x86): na reset vector, običajno fizični naslov 0xFFFFFFF0.

- ARM: na naslov 0x00000000 ali 0xFFFF0000, odvisno od konfiguracije vector table mapping.

**Odgovor**:

    - RFC 3396, RFC 4361 (DUID in identifikacija).

    - 2 = OFFER …

**Izziv:**
- Izvedite vsaj enega od zgornjih napadov. O čem govori RFC 3118 in kako deluje?

**Odgovor**:

    - uporablja deljene ključe ali public-key mehanizme,

V praksi se malo uporablja.

**Kontekst:**
- DHCPv6 je definiran v RFC 3315.           
**Izziv:**
- Poiščite ostale RFC dokumente o DHCPv6 in preverite, kaj piše v njih.

**Odgovor**:
- Primeri: RFC 3315, RFC 3736, RFC 3646, RFC 3319.
- Pokrivajo: DNS options, informacijski način, prefix delegation.

**Kontekst:**
- DHCPv6 uporablja različne tipe sporočil (SOLICIT, ADVERTISE, REQUEST, itd.).      
**Izziv:**
- Kako deluje prepošiljanje zahtev?

**Odgovor**:
- Relay:
    1. prejme sporočilo od odjemalca,
    2. ga zapakira v RELAY-FORW,
    3. posreduje strežniku;
    - strežnik odgovori z RELAY-REPL,
    - relay ga razpakira in posreduje odjemalcu.

**Kontekst:**
- DHCPv6 paket ima obliko msg-type + transaction-id + options.      
**Izziv:**
- Kakšne možnosti (options) obstajajo? Kam so izginila polja iz IPv4? Kaj je DUID?

**Odgovor**:
- Options vključujejo: DNS, domain search, preference, server ID, IA_NA, IA_PD, time parameters.
- IPv4 polja so bila odstranjena (npr. siaddr, giaddr) in nadomeščena z options.
- DUID = DHCP Unique Identifier, identifikator strežnika ali odjemalca.

**Kontekst:**
- Na FreeBSD DHCPv6 uporablja programe dhcp6c, dhcp6s in dhcp6relay, konfiguracije so v /usr/local/etc/dhcp6c.conf in /usr/local/etc/dhcp6s.conf.       
**Izziv:**
- Skonfigurirajte odjemalca in ga poženite. Kaj pomeni dana konfiguracija?

**Odgovor**:
- Opisuje:
    - DUID odjemalca,
    - katere parametre naj zahteva (DNS, prefix),
    - katere vmesnike naj uporablja,
    - kako naj obnavlja konfiguracijo.
- Ob zagonu dhcp6c pridobi nastavitve od strežnika

## Nadzor in upravljanje

**Kontekst:**
- SNMP uporablja drevesno strukturo identifikatorjev (OID). Vsak objekt je predstavljen s številčno potjo od korena do lista (npr. 1.3.6.1.2.1.7 = UDP).        
**Izziv:**
- Kaj se nahaja na drugem in tretjem nivoju drevesa identifikatorjev?

**Odgovor**:
- 1 → iso
- 1.3 → identified organization (org)

**Kontekst:**
- SNMP je definiran v več RFC dokumentih, ki opisujejo različne verzije in razširitve.      
**Izziv:**
- Poiščite RFC dokumente o SNMP in ugotovite razlike med njimi.

**Odgovor**:
- RFC 1157 – SNMPv1: osnovni protokol, brez varnosti.
- RFC 1905/1906/1907 – SNMPv2: nove vrste sporočil, boljša zmogljivost, še vedno brez varnosti.
- RFC 3410–3418 – SNMPv3: avtentikacija, enkripcija, nadzor dostopa.

**Kontekst:**
- SNMP uporablja UDP (vrata 161, 162) in se sooča z omejitvami: velikost paketov, izguba paketov, pomanjkanje potrjevanja dostave.      
**Izziv:**
- Kako SNMPv3 rešuje navedene težave?

**Odgovor**:
- Integriteta: HMAC (SHA-1, MD5).
- Zaupnost: AES enkripcija.
- Ponovno pošiljanje: Z uporabo kombinacije ID pošiljatelja, števila zagonov in časa po zadnjem zagonu. 
    - Ali se **ujema ID pošiljatelja** (engineID)?
    - Je **engineBoots** (št. zagovov) **≥** zadnje znane vredn**osti?
    - Je **engineTime** (čas od zadnjega zagona pošiljatelja) **znotraj dovoljenega časovnega okna**?
- Izguba sporočil: potrjevanje na aplikacijski ravni in varni modeli transporta.

**Kontekst:**
- SNMP uporablja Request ID za ujemanje zahtev in odgovorov ter Error Status za sporočanje napak.       
**Izziv:**
- Poglej različne tipe napak.

**Odgovor**:
- Primeri (SNMPv1):
    - tooBig – odgovor predolg
    - noSuchName – objekt ne obstaja
    - badValue – neveljavna vrednost
    - readOnly – ni dovoljeno pisati
    - genErr – splošna napaka

**Kontekst:**
- SetRequest lahko spreminja konfiguracijo naprav, zato je potrebna zaščita pred nepooblaščenimi akcijami.      
**Izziv:**
- Poišči še 3 primere drugih možnih zlorab protokola SNMP.

**Odgovor**:
- 3 primeri:
    - Branje občutljivih podatkov (npr. routing tabel).
    - Spreminjanje konfiguracij (SetRequest).
    - Napadi z lažnimi trap sporočili.

**Kontekst:**
- SNMPv3 uvaja varnost na osnovi uporabniških imen in avtentikacije.    
**Izziv:**
- Proti kakšnim vdorom omogoča SNMPv3 zaščito? Kako je z napadi DoS in prisluškovanjem?

**Odgovor**:
- Replay napadi: časovni žigi.
- Neavtoriziran dostop: uporabniško ime + HMAC.
- Prisluškovanje: AES enkripcija.
- DoS: ni posebej zaščiten; še vedno ranljiv proti flood napadom.

**Kontekst:**
- SNMPv3 uporablja View-based Access Control Model (VACM), ki določa pravice uporabnikov.   
**Izziv:**
- Kaj je to View-based Access Control Model Configuration MIB?

**Odgovor**:
- VACM je model, ki določa:
    - katere objekte lahko vidi posamezen uporabnik,
    - katere operacije lahko izvaja (read/write/notify),
    - organizira pravila v poglede (views), ki definirajo pravice nad OID drevesom.

**Kontekst:**
- SNMP definira tri modele pretvorbe podatkov med napravami (local → remote format).    
**Izziv:**
- Kakšne so prednosti in slabosti gornjih treh pristopov?

**Odgovor**:
1. Pošiljatelj pretvori v obliko prejemnika
    + Ni potrebe po dodatni obdelavi pri prejemniku
    - Pošiljatelj mora poznati format prejemnika

2. Pošiljatelj pošlje v svoji obliki
    + Najenostavnejše za pošiljatelja
    - Prejemnik mora opraviti pretvorbo

3. Neodvisna (canonical) oblika
    + Najbolj interoperabilno
    + Prilagodljivo za različne platforme
    - Zahteva pretvorbo na obeh straneh

**Kontekst:**
- RFC 789 opisuje velik izpad ARPAneta leta 1980.       
**Izziv:**
- Kako bi lahko izpad preprečili ali pospešili obnovitev z današnjimi orodji za upravljanje omrežja?

**Odgovor**:
- uporaba centraliziranih nadzornih sistemov (npr. SNMPv3, NetFlow),
- avtomatsko odkrivanje anomalij,
- redundanca povezav,
- hitra identifikacija napačnih usmerjevalnih tabel,
- možnost hitrega rollbacka konfiguracij,
- uporaba modernih orodij: Zabbix, Nagios, Grafana, SIEM.

## Stvarni čas in komuniciranje

### **Kontekst** - Real Time

Realni čas pomeni, da mora sistem odgovoriti ali izvesti nalogo v točno določenem časovnem intervalu. Ključni pojmi:

- _čas dospetja_ (arrival time)
- _čas začetka izvajanja_   
- _potreben čas izvajanja_
- _rok zaključka_ (deadline)

Poznamo:
- **Hard real-time** – zamuda NI dovoljena (npr. ABS, letalski sistemi).
- **Soft real-time** – zamuda je dovoljena, vendar poslabša kakovost (npr. videi, VoIP).

#### **Izziv – Ali FreeBSD, Linux in Windows omogočajo real-time?**

**Odgovor (izpitni, jedrnat):**     
Običajne verzije FreeBSD, Linux in Windows **niso hard real-time**, ker ne garantirajo absolutnih časovnih rokov.  
Linux z _PREEMPT_RT_ lahko deluje kot **soft real-time** ali celo blizu hard,  
Windows ima **Windows Embedded/RT**, ki podpira deterministično delovanje,  
FreeBSD pa omogoča **prednostno razporejanje**, vendar ni hard real-time.

### **Kontekst** - Digiralni zvok
- Zvok se vzorči v diskretnih časovnih intervalih → amplituda postane n-bitno število.

#### **Izziv – Audacity**

**Odgovor:**  
Audacity omogoča snemanje, prikaz valovne oblike, filtriranje, rezanje, normalizacijo, izvoz v več formatov (WAV, MP3).

### **Kontekst** - Problem vzorčenja (Nyquist)

#### **Izziv – Zakaj se v filmih kolesa vrtijo nazaj?**

**Odgovor:**  
Gre za **aliasing**. Če je frekvenca vrtenja blizu polovici frekvence snemanja (npr. 24 fps), se faza med vzorci spremeni tako, da vizualno deluje kot gibanje nazaj.

#### **Izziv – Kakšna je frekvenca vzorčenja WAV datotek?**

**Odgovor:**  
Standardni WAV (CD-kvaliteta): **44.1 kHz**, ker pokrije dvakratno najvišjo človeško slišno frekvenco (22 kHz).

#### **Izziv – MP3 kompresija (CLI program v Unixu)**

**Odgovor:**  
Najpogostejši: **LAME** (lame, lame --preset standard input.wav output.mp3).  
Podpira VBR/CBR, psychoacoustic modele, odstranjuje neslišne frekvence.

### **Kontekst** - dogotalizacija slike in videa

#### **Izziv – Tri razsežnosti vektorja slike**

Najpogosteje:

- **RGB** (Red, Green, Blue)  
    Drugi možni: HSV, YUV, CMYK, LAB.
    

#### **Izziv – Standardi JPG, GIF, PNG, BMP**

**Kratek pregled:**

- **JPG** – izgubljiva kompresija, idealen za fotografije.
    
- **PNG** – brezizgubna, podpira prosojnost.
    
- **GIF** – 256 barv, animacije.
    
- **BMP** – nezaželen, velik, brez kompresije.  
    Pretvorba je trivialna, a lahko pride do **izgub pri JPG**.
    

#### **Izziv – Standardne hitrosti slike na sekundo**

- **24 fps** – film
    
- **25 fps** – PAL
    
- **30 fps** – NTSC
    
- **50/60 fps** – šport, igre  
    Različne zaradi zgodovinskih električnih standardov (50 Hz v EU, 60 Hz v ZDA).
    

#### **Izziv – Standardi zapisov filma (formati)**

- **MP4** – trenutno najpogostejši, H.264/H.265
    
- **MKV** – fleksibilen kontejner, več zvočnih stez
    
- **AVI** – starejši, šibka kompresija
    
- **MOV** – Apple  
    Pretvorbe možno delati z _ffmpeg_.
    

### **Kontekst** - NTP (Network Time Protocol)

#### **Izziv – Ostali RFC in Marzullo algoritem**

**Odgovor:**  
NTP RFC-ji opisujejo sinhronizacijo časa, varnostne mehanizme, hierarhijo strežnikov. Marzullo algoritem združuje intervale časa iz različnih virov in izbere najbolj verjeten presek (odpravljanje odstopanj).

#### **Izziv – FreeBSD ntpd**

Ko premakneš čas ročno: ntpd zazna skok in čas postopno popravi (slew), da prepreči nenadne spremembe.

#### **Izziv – NTP na Windows**

Uporablja se **w32time** (`w32tm /resync`). Lahko sinhronizira čas s poljubnim NTP strežnikom.


### **Kontekst** - RTP (Real time Transport Protocol)

#### **Izziv – RFC in TFTP**

RFCji opisujejo časovno označevanje, sinhronizacijo tokov, multicast.  
TFTP RFCji razlagajo preprost transport datotek preko UDP.

#### **Izziv – RTP protokoli po vrstah (zvok, video, tekst)**

Protokoli definirajo,

- kodeke (AAC, H.264, Opus),
    
- clock rate,
    
- timestamp format.  
    Primerjaš RFC-je za posamezne medijske formate.

### **Dodatni podatki**

#### **Izziv – : podatki in CNAME**

RTCP poroča:

- izgube paketov
    
- jitter
    
- zamude
    
- SSRC identifikator
    
- CNAME = canonical name (npr. user@host), stabilna identifikacija pošiljatelja.
    

#### **Izziv – Kako izgleda poročilo o kakovosti prometa?**

Vsebuje: fraction lost, cumulative lost, highest seq. number, jitter, timestampi.

#### **Izziv – var polje pri SR (Sender Report)**

**var = število poročil** v paketu (RC – report count).

#### **Izziv – RTP, FreeBSD in MPlayer**

MPlayer uporablja RTP za predvajanje medijskih tokov in ima podporo za FreeBSD.  
Povezava: FreeBSD podpira RTP-streaming, MPlayer ga zna predvajati.

### **Kontekst** - SRTP (Secure RTP)

#### **Izziv – Izmenjava ključev**

Najpogosteje:

- **SDES** (ključ v SDP) – ni najbolj varen
    
- **DTLS-SRTP** – zelo varen, uporablja DTLS handshake
    
- **ZRTP** – avtonomen protokol z preverjanjem govornega odtisa
    

#### **Izziv – HMAC in funkcija f8**

- **HMAC** = hash-based message authentication (MAC + integriteta).
    
- **f8** = stream cipher mode za šifriranje RTP paketov (podoben OFB), definiran v RFC.
    

### **Kontekst** - RTSP (Reap Time Streaming Protocol)

#### **Izziv – Transport polje**

Pove strežniku,

- ali naj uporablja UDP/TCP,
    
- katere porte,
    
- SSRC in časovne parametre.
    

#### **Izziv – Povezava RTSP ↔ RTP**

V RTSP odgovorih se pojavi **SSRC** (enak kot v RTP), kar poveže signalizacijo z medijskim tokom.

### **Kontekst** - Darwin Straming Server

#### **Izziv – Odjemalec**

Najpogostejši odjemalci: VLC, MPlayer, QuickTime.  
Darwin lahko namestiš na BSD/Linux in dodajaš lastne RTSP/HTTP tokove.


## Razpošiljanje

### **Kontekst** - Multicasting (Osnovni izivi)

#### **Izziv: Zakaj imajo multicast naslovi v usmerjevalnih tabelah več posebnih vnosov?**

**Odgovor:**  
Ker multicast naslovi ne pripadajo nobeni mreži/maski (vedno /32), usmerjevalnik ne more uporabiti klasičnega LPM (Longest Prefix Match).  
Več posebnih vnosov je potrebnih zato, da:

- pokrije več multicast skupin,
    
- obdeluje različna drevesa (source-tree, shared-tree),
    
- ločuje različne protokole (PIM-SM, PIM-DM, DVMRP),
    
- hrani informacije o prejemnikih na različnih vmesnikih.

### **Kontekst** -Multicast konference - kako izgledajo

#### **Izziv: Kako izgleda konferenca z multicast pristopom?**

**Odgovor:**  
Vsak udeleženec, ki govori, pošlje paket na multicast skupino (npr. 239.x.x.x).  
Vsi poslušalci so člani te skupine.  
Če obstaja “konferenčni center”, ta:

- izbere aktivnega govorca,
    
- filtrira nepotrebne pakete,
    
- skrbi za preklapljanje zvočnih/video tokov.  
    Usmerjevalniki posredujejo promet samo na tiste vmesnike, kjer so poslušalci (IGMP/MLD).

### **Kontekst** - Ethernet preslikava multicast naslovov

#### **Izziv: Kaj mora početi usmerjevalnik zaradi nepopolne preslikave (23 bitov)?**

**Odgovor:**  
Ker se 32 IPv4 multicast naslovov preslika v isti Ethernet MAC, mora usmerjevalnik:

- filtrirati pakete po IP-naslovu (ne samo po MAC),
    
- preverjati članstvo IGMP,
    
- preprečevati, da promiscuous multicast MAC ne preplavi omrežja,
    
- razločevati skupine na L3, čeprav so združene na L2.

### **Kontekst** - IGMP

#### **Izziv: Ostali RFC dokumenti o IGMP**

**Odgovor:**  
Ključni:

- RFC 988 – originalni IGMP
    
- RFC 1112 – IGMPv1
    
- RFC 2236 – IGMPv2
    
- RFC 3376 – IGMPv3 (podpora SSM)
    
- RFC 4604 – IGMP/MLD snooping switches

### **Kontekst** - MLD

#### **Izziv: Kaj je Robustness Variable?**

**Odgovor:**  
Parameter, ki določa, kako odporen je protokol na izgube paketov.  
Večja vrednost = več ponovitev IGMP/MLD sporočil = večja zanesljivost, a več prometa.

#### **Izziv: Razlike med IGMP in MLD**

|IGMP|MLD|
|---|---|
|Za IPv4|Za IPv6|
|Deluje nad IP protokolom|Uporablja ICMPv6|
|Skupine: 224.0.0.0/4|FF00::/8|
|V3 podpira SSM|MLDv2 podpira SSM|

#### **Izziv: Sobivanje IGMP in MLD**

**Odgovor:**  
V dual-stack omrežjih delujeta vzporedno:

- IGMP za IPv4 multicast,
    
- MLD za IPv6 multicast.  
    Stikala pogosto izvajajo IGMP/MLD snooping, da filtrirajo L2 multicast promet za oba protokola posebej.

### **Kontekst** - PIM-SM (Protocol Independent Multicast - Sparse Mode)

#### **Izziv: bistvo PIM-SM**

**Odgovor (jedrnato):**  
PIM-SM uporablja shared-tree (_Rendezvous Point, RP_) za registracijo virov, nato pa se lahko preklopi na optimalno source-based drevo (S,G).  
Primeren je za omrežja, kjer je malo oddajnikov in mnogi sprejemniki.

### **Kontekst** - MOSPF

####**Izziv: RFC dokumenti o MOSPF**

**Odgovor:**  
MOSPF definiran v **RFC 1584**.  
Gre za razširitev OSPF: dodani so LSAs, ki opisujejo članstvo v multicast skupinah. Uporablja link-state metode za izračun multicast dreves.

### **Kontekst** - DVMRP

#### **Izziv: Kako deluje (po RFC 1075)?**

**Odgovor:**  
DVMRP temelji na distance-vector metodah. Uporablja RPF (Reverse Path Forwarding) in flood-and-prune, dokler ne ostane optimizirano multicast drevo.  
V začetku je bil glavni protokol v MBONE.


### **Kontekst** - MBONE

#### **Izziv: Kaj piše v RFC 2715 – MBONE?**

**Odgovor:**  
MBONE je virtualno omrežje za testiranje in uporabo multicasta po internetu.  
Zgrajeno je bilo s TUNNELI med usmerjevalniki, ki so lahko prenašali multicast tudi, če omrežje ni bilo multicast-overjeno.

### **Kontekst** - ASM (Any Source Multicast (RFC 1112))

### **Izziv: Kaj je ASM?**

**Odgovor:**  
ASM omogoča, da se sprejemniki pridružijo skupini, ne da bi vedeli, kdo je oddajnik.  
Sprejemnik reče: »Pridruži me skupini G«, nato usmerjevalniki poiščejo vse možne pošiljatelje.  
Alternativa: **SSM (Source Specific Multicast)**, kjer sprejemnik določi tudi oddajnika (S,G).
