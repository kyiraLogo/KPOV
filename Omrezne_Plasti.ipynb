## Omrežna
- Trudi se po najbolših močeh (dostava)
- *USMERJANJE*
- dostava, zaporedje, fregmentacija, izogib. zamašitve

**IPv4:** *(32 bitni naskov*)
- premalo naslovnega prostora (rešitev DHCP)
- time stamp (TTL)
Maska: IP/n prvih n bitov je nespremenjenih. (Kar je zadaj lahko spreminjamo)
št.št.št.št => št = 8 bit (0-255)

- Posebni IP naslovi: (224.0.0.0 - 224.0.0.225)
    - [Posebni_IPv4_naslovi](./Slike/PosebniIPv4.png)
    - [IPv4_posebni_vsi](./Slike/IPv4_posebni_vsi.png)

- Privatni
    - **10.10.0.0/8** (znotraj tega prvih 8 bitov primarno nastavljenih)
    - **172.1 6.0.0/12**
    - **192.168.00/16**
    - **127.0.0.0/8** (local host)
    - **169.254.0.0/16** (link local) ko je kabelj v njega povezan in ne pridobi zunanjega ip naslova, si ga takega sam dodeli (pošle po omrežju da vidi če ima že kdo drug tak ip naslov)
	- Javni
        - **224.0.0.0/4** (multicast) rutatji poskrbijo da se čimbolje porazdeli
        omrežje


**IPv6:** *(128 bitni naslov)*
- **Nima brotcast naslovov zato uporabljamo multi cast**
- [IPv6_packet](./Slike/IPv6_packet.png)
- ne potrebuje NAT (dovolj naslovov)
- št. skokov (Hop limit)
- IPSec ni potrebna
- fregmentacije ni (usmerjevalnik ne sme razbiti paketa na več delov)
št:št:št:št:št:št::št => len(št) = 16, št = 8x, (::) pomeni da je št znaoraj 0 (lahko se pojavi samo enkrat)

- Posebni IPv6: (začetek z "FF")
    - FF02::1 (lokalno omrežje (link local): vsi VMESNIKI)
    - FF02::2 (lokalno omrežje (link local): vsi USMERJEVALNIKI)
    - [IPv6_posebni](./Slike/IPv6_posebni.png)

### IPv6 in IPv4 skupaj:
- **Dvojni sklad:** usmerjevalnik pozna IPv4 in IPv6 (možnost uporaba obeh)
- **Tuneliranje:** IPv6 paket zapakiramo v enega ali več IPv4 paketov kot podatke

## Oprema
- **Usmerjevalnik:** *(router)*
    - omrežna plast
    - NAT, DHCP, fire wall
    - usmerjevalne tabele (naslov mreže : naprava => naslov)
- **Bridge:**
    - povezavna plast
    - starejša verzija stikala
    - dva LAN segmenta
    - prvi “pametni” naslednik Huba
- **Stikalo:** *(switch)*
    - povezavna plast
    - tabela za preklapljanje
    - filtriranje
    - odkrivanje omrezja
- **Povezovališče:** *(hub)*
    - fizična plast


## Entitetni par
Par Entitet, ki se pogovarjajo med seboj na isti plasti med različnimi napravami.
- Aplikaciska,
- Transportna
- Omrežna
- Povezavna
Npr. *(Aplikaciska) Datotečni sistem <--> Program

- **Full duplex:**
    - istočasno lahko pošiljata in prejemata
    - Npr. *(Transportna) A <-Sporočilo-> B
- **Half duplex:**
    - entitetni par lahko deluje samo v eno smer hkrati
    - Npr. *(Omrežna) A -->  <-- B
