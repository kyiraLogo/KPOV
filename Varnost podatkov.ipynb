## Omrežna varnost

Varnostna koncepta:
- **Skrivanje podatkov**
- **Celovitost podatkov**

## Zagotavljanje varnosti
- **Nadzor:**
    - zbiranje podatkov o delovanju, uporabi, dnevnik
- **Upravljanje:**
    - ukrelanje na podlagi zbranih podatkov
    - diagnostika
    - administracija
- **Sistematičnost:**
    - imenik, seznami in kazala
    - SNMP
    - poslovna pravila
- **Načrtovanje:**
    - zmogljivosti
    - razvoj
    - testiranje in uvajanje
- **Razpršenost zaščite:**
    - integracija povezav, virov, vsebine, uporabnikov, sporočil (povezovanje delov v celoto)

## Elementi varne komunikacije
- Zaupnost
- Overovljenje (avtorizacija)
- Razpoložlivost in nadzor dostopa
- Integriteta sporočila
- Onemogočanje zanikanja

**Praksa:**
- fire wall, zaznavanje udorov
- varnost na plasteh

## Napadi
- Prisluškovanje
- Ponarejanje
- Kraja identitete
- Prevzem povezave
- Onemogočanje nudenja storitve:
    - *TCP*: Pošiljamo samo *sim* pakete (samo 3 way hand shake)
    - *UDP*: Pošljemo veliko IP paketov, ki so dovol veliki, da jih strežnik razkosa na fregmente. Nato pa teh fragmentov nikoli ne dokončamo.
